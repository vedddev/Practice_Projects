```python
# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import numpy as np
import pandas as pd
import string
import nltk
import pickle

from nltk.tokenize import word_tokenize
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score, classification_report


# ============================================================
# 2. DOWNLOAD NLTK RESOURCES
# ============================================================

nltk.download('punkt')
nltk.download('punkt_tab')


# ============================================================
# 3. LOAD DATASET
# ============================================================

train_df = pd.read_csv(
    'redmi6.csv',
    encoding='latin1'
)

print("Dataset shape:", train_df.shape)
print("Columns:", train_df.columns.tolist())

print("\nMissing values:")
print(train_df.isnull().sum())


# ============================================================
# 4. EXTRACT NUMERIC RATING
# ============================================================

rates = []

for i in train_df['Rating']:
    rates.append(i[0:3])

rates = pd.to_numeric(rates)

train_df["rates"] = rates

print("\nRatings:")
print(train_df["rates"].value_counts().sort_index())


# ============================================================
# 5. CREATE SENTIMENT LABEL
# ============================================================

def Convert(rating):

    if rating >= 4:
        return "positive"

    elif rating == 3:
        return "neutral"

    else:
        return "negative"


y = train_df['rates'].apply(Convert)

X = train_df['Comments']


# ============================================================
# 6. CHECK SENTIMENT DISTRIBUTION
# ============================================================

print("\nSentiment counts:")
print(y.value_counts())

print("\nSentiment percentages:")
print(y.value_counts(normalize=True) * 100)


# ============================================================
# 7. TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    stratify=y,
    test_size=0.2,
    random_state=42
)

print("\nTraining samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])


# ============================================================
# 8. NLP PREPROCESSING FUNCTION
# ============================================================

def preprocess_text(text):

    tokens = word_tokenize(
        text.lower()
    )

    # Remove punctuation
    tokens = [
        word
        for word in tokens
        if word not in string.punctuation
    ]

    return " ".join(tokens)


# ============================================================
# 9. PREPROCESS TRAINING AND TEST DATA
# ============================================================

X_train_clean = X_train.apply(
    preprocess_text
)

X_test_clean = X_test.apply(
    preprocess_text
)

print("\nExample original:")
print(X_train.iloc[0])

print("\nExample cleaned:")
print(X_train_clean.iloc[0])


# ============================================================
# 10. TF-IDF
#     Unigrams + Bigrams
# ============================================================

tf_idf = TfidfVectorizer(
    lowercase=True,
    stop_words='english',
    ngram_range=(1, 2)
)

X_train_tfidf = tf_idf.fit_transform(
    X_train_clean
)

X_test_tfidf = tf_idf.transform(
    X_test_clean
)

print("\nTF-IDF training shape:")
print(X_train_tfidf.shape)

print("\nTF-IDF testing shape:")
print(X_test_tfidf.shape)


# ============================================================
# 11. LOGISTIC REGRESSION
# ============================================================

Lg = LogisticRegression(
    class_weight='balanced',
    max_iter=1000,
    random_state=42
)

Lg.fit(
    X_train_tfidf,
    y_train
)


# ============================================================
# 12. MODEL EVALUATION
# ============================================================

y_pred = Lg.predict(
    X_test_tfidf
)

f1 = f1_score(
    y_test,
    y_pred,
    average='weighted'
)

print("\n================================")
print("MODEL PERFORMANCE")
print("================================")

print("\nWeighted F1 Score:")
print(f1)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred
    )
)


# ============================================================
# 13. TEST CUSTOM REVIEWS
# ============================================================

test_reviews = [

    "The battery backup is terrible and the phone heats up during normal use.",

    "The phone is very slow and hangs frequently. I am extremely disappointed.",

    "The camera quality is poor and the battery drains very quickly.",

    "Amazing phone. The battery lasts all day and the performance is excellent."
]


print("\n================================")
print("CUSTOM REVIEW TESTING")
print("================================")


for review in test_reviews:

    # Same preprocessing used during training
    clean_review = preprocess_text(
        review
    )

    # Convert to TF-IDF
    review_tfidf = tf_idf.transform(
        [clean_review]
    )

    # Prediction
    prediction = Lg.predict(
        review_tfidf
    )[0]

    # Probability
    probabilities = Lg.predict_proba(
        review_tfidf
    )[0]

    print("\n--------------------------------")
    print("Review:")
    print(review)

    print("\nPrediction:")
    print(prediction)

    print("\nProbabilities:")

    for class_name, probability in zip(
        Lg.classes_,
        probabilities
    ):
        print(
            f"{class_name}: {probability:.4f}"
        )


# ============================================================
# 14. SAVE TRAINED MODEL
# ============================================================

with open(
    'sentiment_model.pkl',
    'wb'
) as f:

    pickle.dump(
        Lg,
        f
    )


# ============================================================
# 15. SAVE TF-IDF VECTORIZER
# ============================================================

with open(
    'tfidf_vectorizer.pkl',
    'wb'
) as f:

    pickle.dump(
        tf_idf,
        f
    )


print("\n================================")
print("FILES SAVED SUCCESSFULLY")
print("================================")

print("sentiment_model.pkl")
print("tfidf_vectorizer.pkl")
```

```python
feature_names = tf_idf.get_feature_names_out()

print("Number of features:", len(feature_names))

for class_name, coefficients in zip(
    Lg.classes_,
    Lg.coef_
):

    print("\n==============================")
    print("CLASS:", class_name)
    print("==============================")

    top_indices = coefficients.argsort()[-20:][::-1]

    for index in top_indices:
        print(
            f"{feature_names[index]:30s} "
            f"{coefficients[index]:.4f}"
        )
```
