```python
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
import string 
import contractions 
def preprocess(text):     
    text=contractions.fix(text)     
    tokens=word_tokenize(text.lower())     
    tokens=[word for word in tokens if word not in string.punctuation]     
    return tokens
```
